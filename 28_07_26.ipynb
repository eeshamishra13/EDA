HANDLING MISSING VALUES

```python
# Importing required libraries

import numpy as np
import pandas as pd
```

```python
# Creating a DataFrame

data = np.arange(15, 30).reshape(5, 3)

dfx = pd.DataFrame(data,
                   index=['apple', 'banana', 'kiwi', 'grapes', 'mango'],
                   columns=['store1', 'store2', 'store3'])

dfx
```

```python
# Adding missing values to the DataFrame

dfx['store4'] = np.nan

dfx.loc['watermelon'] = np.arange(15, 19)

dfx.loc['oranges'] = np.nan

dfx['store5'] = np.nan

dfx.loc['apple', 'store4'] = 20.

dfx
```

```python
# Checking for missing values

dfx.isnull()
```

```python
# Checking for non-missing values

dfx.notnull()
```

```python
# Counting missing values in each column

dfx.isnull().sum()
```

```python
# Counting total missing values in the DataFrame

dfx.isnull().sum().sum()
```

```python
# Counting non-missing values in each column

dfx.count()
```

```python
# Displaying non-missing values from store4

dfx.store4[dfx.store4.notnull()]
```

```python
# Dropping missing values from store4

dfx.store4.dropna()
```

```python
# Dropping rows containing missing values

dfx.dropna()
```

```python
# Dropping rows where all values are missing

dfx.dropna(how='all')
```

```python
# Dropping columns where all values are missing

dfx.dropna(how='all', axis=1)
```

```python
# Dropping columns having less than 5 non-missing values

dfx.dropna(thresh=5, axis=1)
```

```python
# Performing mathematical operations with NaN values

ar1 = np.array([100, 200, np.nan, 300])

ser1 = pd.Series(ar1)

ar1.mean(), ser1.mean()
```

```python
# Calculating the sum of store4

ser2 = dfx.store4

ser2.sum()
```

```python
# Calculating the mean of store4

ser2.mean()
```

```python
# Calculating cumulative sum

ser2.cumsum()
```

```python
# Filling missing values with 0

filledDf = dfx.fillna(0)

filledDf
```

```python
# Comparing mean before filling missing values

dfx.mean()
```

```python
# Comparing mean after filling missing values

filledDf.mean()
```

```python
# Forward filling missing values

dfx.store4.ffill()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create a heatmap to visualize missing values
plt.figure(figsize=(10, 6))
sns.heatmap(dfx.isnull(), cbar=False, cmap='viridis')
plt.title('Heatmap of Missing Values in dfx')
plt.xlabel('Columns')
plt.ylabel('Rows')
plt.show()
```

```python
# Calculate the percentage of missing values for each column
missing_percentage = (dfx.isnull().sum() / len(dfx)) * 100

print("Percentage of missing values per column:")
display(missing_percentage)
```

IMPUTATION TECHNIQUES

```python
# Importing required libraries

import numpy as np
import pandas as pd

from sklearn.experimental import enable_iterative_imputer

from sklearn.impute import IterativeImputer, SimpleImputer
```

```python
# Creating the dataset

df = pd.DataFrame({

    "Age": [25, np.nan, 30, 45, 22],

    "Salary": [50000, 60000, np.nan, 80000, 45000],

    "Category": ["A", "B", "A", np.nan, "B"]

})
```

```python
# Mean Imputation

mean_imputer = SimpleImputer(strategy="mean")

df["Age_Mean"] = mean_imputer.fit_transform(df[["Age"]])
```

```python
# Median Imputation

median_imputer = SimpleImputer(strategy="median")

df["Salary_Median"] = median_imputer.fit_transform(df[["Salary"]])
```

```python
# Mode Imputation

mode_imputer = SimpleImputer(strategy="most_frequent")

df["Category_Mode"] = mode_imputer.fit_transform(df[["Category"]]).flatten()
```

```python
# MICE (Iterative) Imputation

num_cols = ["Age", "Salary"]

mice_imputer = IterativeImputer(max_iter=10, random_state=42)

df_mice = df[num_cols].copy()

df_mice.iloc[:, :] = mice_imputer.fit_transform(df_mice)

df["Age_MICE"] = df_mice["Age"]

df["Salary_MICE"] = df_mice["Salary"]

print(df)
```

```python
# Replacing missing categorical values with a constant value

constant_imputer = SimpleImputer(strategy="constant", fill_value="Unknown")

df["Category_Constant"] = constant_imputer.fit_transform(df[["Category"]]).flatten()

print(df)
```
